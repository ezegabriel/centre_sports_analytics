import plotly.graph_objects as go
import numpy as np

# Create figure
fig = go.Figure()

# Add traces, one for each slider step
for step in np.arange(0, 5, 0.1):
    fig.add_trace(
        go.Scatter(
            visible=False,
            line=dict(color="#00CED1", width=6),
            name="𝜈 = " + str(step),
            x=np.arange(0, 10, 0.01),
            y=np.sin(step * np.arange(0, 10, 0.01))))

# Make 10th trace visible
fig.data[10].visible = True

# Create and add slider
steps = []
for i in range(len(fig.data)):
    step = dict(
        method="update",
        args=[{"visible": [False] * len(fig.data)},
              {"title": "Slider switched to step: " + str(i)}],  # layout attribute
    )
    step["args"][0]["visible"][i] = True  # Toggle i'th trace to "visible"
    steps.append(step)

sliders = [dict(
    active=10,
    currentvalue={"prefix": "Frequency: "},
    pad={"t": 50},
    steps=steps
)]

fig.update_layout(
    sliders=sliders
)

fig.show()

```python
df = pd.read_csv('/Users/ericmerdian/Desktop/Desktop/A_Summer_Research/AllTeams/submission.csv')
win_prob_round = []

for index, row in df.iterrows():

    rounded = round(float(row['WinProb']) *20)/20 # rounds to nearest .5

    win_prob_round.append(rounded)

df['WinProbRounded'] = win_prob_round
df

```

```python
probs = []
observed_probs = []

quarters = ['q_1', 'q_2', 'q_3', 'q_4']
#print(df[(df['WinProbRounded'] == .9) & (df['Outcome'] == 1)])
i = 1
# for df in q_1, q_2, q_3, q_4:

plt.clf()
for prob in range(0,105,5):
    prob = prob/100
    print('prob ',prob)
    num_observed = len(df[df['WinProbRounded'] == prob])
    print('num observed, ', num_observed)
    num_observed_win = len(df[(df['WinProbRounded'] == prob) & (df['Outcome'] == 1)])
    print('num observed win, ', num_observed_win)
    observed_prob = num_observed_win / num_observed
    print('obs prob, ',observed_prob)
    probs.append(prob)
    observed_probs.append(observed_prob)
colors = ['r', 'b','y','p']
plt.scatter(probs,observed_probs, c=colors[i])
plt.xlim([0,1])
plt.ylim([0,1])
#plt.title("All Data", fontsize=20)
plt.xlabel('Predicted Probability', fontsize=15)
plt.ylabel('Actual Probability', fontsize=15)
x = np.linspace(0,1,100)
plt.plot(x, x, '-r')
plt.show()
i+=1


    
# print(probs)
# print(observed_probs)
    
```
