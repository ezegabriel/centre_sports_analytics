## What is the problem we are trying to solve?

We were tasked with the creation of a win probability model for Centre by Dr. Heath during the summer of 2021. To do this, we first needed to spend an extensive amount of time gathering and familiarizing ourselves with the data for the project. Initially, this meant simply scraping the data from our own centrecolonels.com website for 13 different men's and women's sports. However, we came to find after spending the summer scraping all of that data that we would need to have the data of the other colleges in the conference as well. So, we went about modifying our code to work with the other six colleges in the conference that allowed web scraping. From there, Shelby and I went down two parallel paths from one another with one working on basketball and the other on lacrosse.

## What data were we trying to gather?

The data that we were looking to scrape was publicly available for anyone to see on the web and we were tasked with writings scripts that could convert the information into a usable format. For us, that meant converting the HTML tables on the websites into csv files. The data itself was being hosted across each of the SAA's college websites. On these sites, we were to scrape all the play by play data we could for each of the men's and women's sports.

## What challenges did we face?

Since every sport has its own quirks and bits that make it unique, some of the webpages were formatted differently by sport, while others were simply missing information for the opposing team. We took all this into consideration when we scraped the data to make sure we avoided the problem of GIGO. What this meant was making our scraping tools ensure the full information was available before using it.

## What kind of model did we choose and why?

From the project's very inception, logistic regression was the model of choice to use. The reason was twofold: Not only is logistic regression a very simply model to build and maintain, but it also fit perfectly with the output we were looking for. Since we wanted to build a win probability model that would progress over the course of the game rather than simply trying to predict a win/loss, the predict probability form of logistic regression was a perfect match for what we were looking to build.

## How well did our model do overall?

Logistic regression models like these are a little tricky to say how effective they are, especially for predicting the outcomes of sports. By nature, there is a good amount of uncertainty in the outcome of a game, otherwise they wouldn't be as fun to watch. What we were interested in seeing is how well the models compared when predicting the probability of winning with the actual probability of winning. First, we wanted to look at the data from a big picture standpoint to see how well it performed. From the graph we see below, the data generally follows the y=x line, but generally tends to underpredict.

Looking further into a quarter-by-quarter breakdown, we can see that the first and second quarters generally tend to follow the y=x line, but the third and fourth quarters are somewhat sigmoidal in shape. We can't necessarily say why exactly this is happening, but our inclination is that by the fourth quarter, the vast majority of games are already decided, so there is less data available around the 50/50 probability.

## Next Steps

Looking into the future, we're interested in expanding the horizons of the model beyond basketball and lacrosse. Additionally, we focused on data pulled from the 2017 and 2018 seasons for both sports in the men's division and would be interested in seeing how including more historical data for our training set might alter the strength of our models, especially when it comes to understanding our model evaluation above.
